```python
import pandas as pd
import pyarrow.parquet as pq

# Load the parquet file using pyarrow directly
table = pq.read_table('data/train-00000-of-00001-e97e83ad56524b9c.parquet')
df = table.to_pandas()

# Show the head of the dataframe
df.head()
```

```python
import os
from PIL import Image
import io
from collections import defaultdict

# Create data folder and train/test subfolders if they don't exist
os.makedirs('data', exist_ok=True)
os.makedirs('data/train', exist_ok=True)
os.makedirs('data/test', exist_ok=True)

# Count images per label to manage train/test split
label_counts = defaultdict(int)

# Process each row in the dataframe
for index, row in df.iterrows():
    # Get the image bytes and label
    image_bytes = row['image']['bytes']
    label = row['text']
    
    # Determine if this should go to test (first 5 per label) or train
    if label_counts[label] < 5:
        base_folder = 'data/test'
    else:
        base_folder = 'data/train'
    # Create label folder inside base folder
    label_folder = os.path.join(base_folder, label)
    os.makedirs(label_folder, exist_ok=True)
    
    # Convert bytes to PIL Image
    image = Image.open(io.BytesIO(image_bytes))
    
    # Save image with index as filename
    image_path = os.path.join(label_folder, f'{index}.jpg')
    image.save(image_path)
    
    # Increment count for this label
    label_counts[label] += 1
    
    print(f"Saved image {index} to {image_path}")
```

```python
import requests
import base64

def call_search_api(image_path):
    # Read and encode image to base64
    with open(image_path, 'rb') as image_file:
        image_data = image_file.read()
        base64_string = base64.b64encode(image_data).decode('utf-8')
    
    # API endpoint and headers
    url = 'http://13.212.33.57:8001/search'
    headers = {
        'accept': 'application/json',
        'Content-Type': 'application/json'
    }
    
    # Request payload
    payload = {
        'image': base64_string
    }
    print(base64_string)
    # Make POST request
    response = requests.post(url, headers=headers, json=payload)
    
    return response.json()

# Example usage
path = "data/test/ca sĩ Akira Phan/0.jpg"
result = call_search_api(path)
print(result)
```
